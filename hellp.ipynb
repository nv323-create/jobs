```python
import pandas as pd  # tables

numbers = [1, 2, 3, 4, 5]  # the numbers 1 to 5
squares = [n ** 2 for n in numbers]  # square each number
table = pd.DataFrame({"Number": numbers, "Square": squares})  # build the table
table  # show the table
```

```python
import matplotlib.pyplot as plt  # charts

plt.bar(table["Number"], table["Square"])  # draw the bars
plt.xlabel("Number")  # label the x axis
plt.ylabel("Square")  # label the y axis
plt.title("Squares of 1 to 5")  # add a title
plt.show()  # display the chart
```
